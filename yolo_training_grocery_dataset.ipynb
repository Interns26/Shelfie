```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")
```

```python
results = model.train(
    data=dataset.location + "/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    workers=2
)
```

```python
metrics = model.val()
```
